**Calculationns for statistics**

```python
%%writefile calculate.cpp
#include <iostream>
#include <string>
#include <cmath> // Required for pow() and sqrt()

using namespace std;

int main() {
    int a;
    int b;

    // Getting user inputs
    cout << "Enter value a: ";
    cin >> a;
    cout << "Enter value b: ";
    cin >> b;

    // Math Operations
    int multiply = a * b;
    int add = a + b;
    int subtract = a - b;
    int reminder = a % b;


    int div = a / b;
    double fdivision = (double)a / b;

    // Square Roots
    double sqrt_a = sqrt(a);
    double sqrt_b = sqrt(b);

    // Printing Outputs beautifully
    cout << "\n--- RESULTS ---" << endl;
    cout << "Addition (a + b):          " << add << endl;
    cout << "Subtraction (a - b):       " << subtract << endl;
    cout << "Multiplication (a * b):    " << multiply << endl;
    cout << "Integer Division (a / b):  " << div << endl;
    cout << "Decimal Division (a / b):  " << fdivision << endl;
    cout << "Remainder/Modulus (a % b): " << reminder << endl;
    cout << "Square root of a:          " << sqrt_a << endl;
    cout << "Square root of b:          " << sqrt_b << endl;

    // power where we can directly bring ouput
    cout<< "power of a         "<<pow(a,b)<<endl;

    return 0;
}
```

```python
!g++ calculate.cpp -o my_program && ./my_program
```

```python
%%writefile if_else_and_or.cpp
#include<iostream>
using namespace std;
int main(){
int a = 2;
int b = 3;
int c;
cin>> c;
cout<<"a="<<a<<endl;
cout<<"b="<<b<<endl;
cout<<"c="<<c<<endl;
if ((a || b) < c){
cout<<"small c"<<endl;
}
else{
  cout<<"big c"<<endl;
}

cout<<"---------------------------------------------------"<<endl;
if ((a && b) < c){// a common trap in c++ when using and operator it cahnges any number to 1 or 0 so c is always going to be high
cout<<"small c"<<endl;
}
else{
  cout<<"big c"<<endl;
}
return 0;


}
```
