#What is a Stack?
A stack is a linear data structure that follows the LIFO (Last-In, First-Out) principle. Think of it like a stack of plates in a cafeteria: the last plate you place on top is the first one you have to take off.

In C++, the Standard Template Library (STL) provides a built-in std::stack container that allows you to store elements and manipulate them easily.

Core Operations
push(): Adds an element to the top of the stack.

pop(): Removes the element from the top of the stack.

top(): Returns the element at the top without removing it.

empty(): Checks if the stack is empty.

size(): Returns the number of elements in the stack.

Code Example
Here is a simple, clean example showing how to create a stack of integers, add some numbers, look at the top one, and remove them.

C++
#include <iostream>
#include <stack> // Required header for using stack

int main() {
    // 1. Create a stack of integers
    std::stack<int> myStack;

    // 2. Push elements onto the stack
    myStack.push(10);
    myStack.push(20);
    myStack.push(30); // This is now at the top

    // 3. Check the top element
    std::cout << "Top element: " << myStack.top() << std::endl; // Outputs 30

    // 4. Remove the top element
    myStack.pop(); // Removes 30; 20 is now the new top

    std::cout << "New top element after pop: " << myStack.top() << std::endl; // Outputs 20

    // 5. Check the size of the stack
    std::cout << "Stack size: " << myStack.size() << std::endl; // Outputs 2

    // 6. Check if it's empty
    if (myStack.empty()) {
        std::cout << "The stack is empty." << std::endl;
    } else {
        std::cout << "The stack is not empty." << std::endl;
    }

    return 0;
}
Output
Plaintext
Top element: 30
New top element after pop: 20
Stack size: 2
The stack is not empty.

#What is a Heap?
In C++ (and computer science in general), the word Heap has two different meanings depending on the context:

The Data Structure (Binary Heap): A specialized tree-based data structure that satisfies the "heap property." It is commonly used to implement priority queues.

The Memory Region (Heap Memory): A region of your computer's memory used for dynamic memory allocation (where you manually allocate memory using new and free it using delete).

Since you just asked about the stack data structure, let's focus on the Heap Data Structure.

The Heap Data Structure (Max-Heap vs Min-Heap)
A heap is a complete binary tree. There are two types:

Max-Heap: The root node is the largest element, and every parent node is greater than or equal to its children.

Min-Heap: The root node is the smallest element, and every parent node is less than or equal to its children.

In C++, you can easily create a heap using std::priority_queue from the <queue> header. By default, C++ creates a Max-Heap.

Code Example (Max-Heap)
Here is a short example demonstrating how a Max-Heap automatically keeps the largest element at the top.

C++
#include <iostream>
#include <queue> // Required header for priority_queue (heap)

int main() {
    // 1. Create a Max-Heap (Priority Queue)
    std::priority_queue<int> maxHeap;

    // 2. Insert elements in random order
    maxHeap.push(10);
    maxHeap.push(30);
    maxHeap.push(20);
    maxHeap.push(5);

    // 3. The top element will always be the largest (30)
    std::cout << "Top element (largest): " << maxHeap.top() << std::endl;

    // 4. Remove the top element
    maxHeap.pop(); // Removes 30; 20 becomes the new top

    std::cout << "New top element after pop: " << maxHeap.top() << std::endl;

    // 5. Print and empty the remaining heap
    std::cout << "Remaining elements in order: ";
    while (!maxHeap.empty()) {
        std::cout << maxHeap.top() << " ";
        maxHeap.pop();
    }
    std::cout << std::endl;

    return 0;
}
Output
Plaintext
Top element (largest): 30
New top element after pop: 20
Remaining elements in order: 20 10 5

# Operation overloading

````python
%%writefile OPERATOROBJ.cpp
#include <iostream>
using namespace std;

class Number
{
public:

    int value;                    // Our object has one integer

    Number add(Number other)      // A normal function called add
    {
        Number result;            // Create a new Number object

        result.value = value + other.value;
        // Add the values of the two Number objects

        return result;            // Return the resulting Number
    }
};

int main()
{
    Number n1;                    // Create Number object
    n1.value = 10;                // Put 10 inside it

    Number n2;                    // Create another Number object
    n2.value = 20;                // Put 20 inside it

    Number n3 = n1.add(n2);       // Call the add() function

    cout << n3.value << endl;     // Print 30
}
````

````python
%%writefile OPERATOROBJ.cpp
#include <iostream>
using namespace std;

class Number
{
public:

    int value;                         // The Number object stores an integer

    Number(int v = 0)                  // Constructor
    {
        value = v;                     // Store v inside value
    }

    Number operator+(Number other)     // Define what + means
    {
        Number result;                 // Create a new Number object

        result.value = value + other.value;
        // Add the value of this object
        // to the value of the other object

        return result;                 // Return the new Number object
    }
};

int main()
{
    Number n1(10);                     // n1.value = 10

    Number n2(20);                     // n2.value = 20

    Number n3 = n1 + n2;               // Use our overloaded + operator

    cout << n3.value << endl;          // Print 30
}
````

````python
%%writefile copy_assignment_operator.cpp

#include <iostream>
using namespace std;

class Number
{
public:
    int value;

    // Normal constructor
    Number(int v)
    {
        value = v;
    }

    // Copy assignment operator
    Number& operator=(const Number& other)
    {
        value = other.value;

        return *this;
    }
};

int main()
{
    Number n1(10);
    Number n2(20);

    cout << "Before assignment:" << endl;
    cout << "n1 = " << n1.value << endl;
    cout << "n2 = " << n2.value << endl;

    // Copy assignment
    n2 = n1;

    cout << "After assignment:" << endl;
    cout << "n1 = " << n1.value << endl;
    cout << "n2 = " << n2.value << endl;

    return 0;
}
````

````python
%%writefile template_specialization.cpp
#include <iostream>
using namespace std;

// GENERIC version
template <typename T>
void show(T value) {
    cout << "Generic: " << value << endl;
}

// SPECIALIZED version — only for bool
template <>
void show(bool value) {
    cout << "Bool: " << (value ? "true" : "false") << endl;
}

int main() {
    show(10);       // Generic: 10
    show(3.14);     // Generic: 3.14
    show(true);     // Bool: true   <- uses the specialized veriosn

    return 0;
}
````

#array


````python
%%writefile array.cpp
#include <iostream>
#include <array>
using namespace std;

int main() {

    array<int, 5> a = {10, 20, 30, 40, 50};


    // =========================
    // ACCESS
    // =========================

    cout << "First element -> " << a.front() << endl;
    cout << "Last element  -> " << a.back() << endl;

    cout << "Element at index 2 -> " << a[2] << endl;

    cout << "Element at index 3 -> " << a.at(3) << endl;


    // =========================
    // CHANGE
    // =========================

    cout << "Before change -> " << a[1] << endl;

    a[1] = 25;

    cout << "After change  -> " << a[1] << endl;


    // Change using at()
    cout << "Before change -> " << a.at(2) << endl;

    a.at(2) = 35;

    cout << "After change  -> " << a.at(2) << endl;


    // Change first
    cout << "First before -> " << a.front() << endl;

    a.front() = 5;

    cout << "First after  -> " << a.front() << endl;


    // Change last
    cout << "Last before -> " << a.back() << endl;

    a.back() = 55;

    cout << "Last after  -> " << a.back() << endl;


    // =========================
    // SIZE
    // =========================

    cout << "Size -> " << a.size() << endl;


    // =========================
    // EMPTY
    // =========================

    cout << "Empty -> " << a.empty() << endl;


    // =========================
    // FILL
    // =========================

    a.fill(100);

    cout << "After fill -> ";

    for (int x : a) {
        cout << x << " ";
    }

    cout << endl;


    // =========================
    // ITERATOR
    // =========================

    auto it = a.begin();

    cout << "begin() -> " << *it << endl;

    ++it;

    cout << "After ++it -> " << *it << endl;


    // =========================
    // SORT
    // =========================

    a = {50, 10, 40, 20, 30};

    cout << "Before sort -> ";

    for (int x : a) {
        cout << x << " ";
    }

    cout << endl;

    sort(a.begin(), a.end());

    cout << "After sort -> ";

    for (int x : a) {
        cout << x << " ";
    }

    cout << endl;


    // =========================
    // REVERSE
    // =========================

    reverse(a.begin(), a.end());

    cout << "After reverse -> ";

    for (int x : a) {
        cout << x << " ";
    }

    cout << endl;


    return 0;
}
````

````python
!!g++ array.cpp -o output&& ./output
````

#vector


````python
%%writefile vector.cpp
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;

int main() {

    vector<int> v = {10, 20, 30};


    // =========================
    // ACCESS
    // =========================

    cout << "First element -> " << v.front() << endl;
    cout << "Last element  -> " << v.back() << endl;

    cout << "Index 1 -> " << v[1] << endl;
    cout << "Index 2 -> " << v.at(2) << endl;


    // =========================
    // CHANGE
    // =========================

    cout << "Before change -> " << v[1] << endl;

    v[1] = 25;

    cout << "After change  -> " << v[1] << endl;


    // Change using at()
    cout << "Before change -> " << v.at(2) << endl;

    v.at(2) = 35;

    cout << "After change  -> " << v.at(2) << endl;


    // Change first
    cout << "First before -> " << v.front() << endl;

    v.front() = 5;

    cout << "First after  -> " << v.front() << endl;


    // Change last
    cout << "Last before -> " << v.back() << endl;

    v.back() = 55;

    cout << "Last after  -> " << v.back() << endl;


    // =========================
    // ADD
    // =========================

    v.push_back(60);

    cout << "push_back -> added: " << v.back() << endl;


    // insert
    auto it = v.begin();
    ++it;

    cout << "insert -> inserting 15 before: " << *it << endl;

    v.insert(it, 15);


    // =========================
    // REMOVE
    // =========================

    cout << "pop_back -> removing: " << v.back() << endl;

    v.pop_back();

    cout << "New back -> " << v.back() << endl;


    // erase
    it = v.begin();
    ++it;

    cout << "erase -> removing: " << *it << endl;

    v.erase(it);


    // =========================
    // SIZE
    // =========================

    cout << "Size -> " << v.size() << endl;


    // =========================
    // EMPTY
    // =========================

    cout << "Empty -> " << v.empty() << endl;


    // =========================
    // ITERATORS
    // =========================

    it = v.begin();

    cout << "begin() -> " << *it << endl;

    ++it;

    cout << "After ++it -> " << *it << endl;


    // =========================
    // SORT
    // =========================

    v = {50, 10, 40, 20, 30};

    cout << "Before sort -> ";

    for (int x : v) {
        cout << x << " ";
    }

    cout << endl;

    sort(v.begin(), v.end());

    cout << "After sort -> ";

    for (int x : v) {
        cout << x << " ";
    }

    cout << endl;


    // =========================
    // REVERSE
    // =========================

    reverse(v.begin(), v.end());

    cout << "After reverse -> ";

    for (int x : v) {
        cout << x << " ";
    }

    cout << endl;


    // =========================
    // FILL
    // =========================

    v.assign(5, 100);

    cout << "After assign -> ";

    for (int x : v) {
        cout << x << " ";
    }

    cout << endl;


    // =========================
    // CLEAR
    // =========================

    v.clear();

    cout << "Size after clear -> " << v.size() << endl;
    cout << "Empty after clear -> " << v.empty() << endl;


    return 0;
}
````

````python
%%writefile list.cpp
#include <iostream>
#include <list>
using namespace std;
int main() {
    list<int> l = {10, 20, 30};
    l.push_back(40);
    cout << "push_back -> added: " << l.back() << endl;
    l.push_front(5);
    cout << "push_front -> added: " << l.front() << endl;
    auto it = l.begin();
    ++it;
    cout << "insert -> inserting 15 before: " << *it << endl;
    l.insert(it, 15);
    cout << "front -> " << l.front() << endl;
    cout << "back  -> " << l.back() << endl;
    cout << "size  -> " << l.size() << endl;
    cout << "empty -> " << l.empty() << endl;
    it = l.begin();
    ++it;
    cout << "Before change -> " << *it << endl;
    *it = 25;
    cout << "After change  -> " << *it << endl;
    cout << "First before -> " << l.front() << endl;
    l.front() = 100;
    cout << "First after  -> " << l.front() << endl;
    cout << "Last before -> " << l.back() << endl;
    l.back() = 200;
    cout << "Last after  -> " << l.back() << endl;
    cout << "pop_front -> removing: " << l.front() << endl;
    l.pop_front();
    cout << "New front -> " << l.front() << endl;
    cout << "pop_back -> removing: " << l.back() << endl;
    l.pop_back();
    cout << "New back -> " << l.back() << endl;
    cout << "remove -> removing value: 30" << endl;
    l.remove(30);
    it = l.begin();
    cout << "erase -> removing value: " << *it << endl;
    l.erase(it);
    l.push_back(50);
    l.push_back(10);
    l.push_back(40);
    cout << "sort -> sorting list" << endl;
    l.sort();
    cout << "First after sort -> " << l.front() << endl;
    cout << "reverse -> reversing list" << endl;
    l.reverse();
    cout << "First after reverse -> " << l.front() << endl;
    l.push_back(50);
    l.push_back(50);
    cout << "unique -> removing consecutive duplicates" << endl;
    l.unique();
    cout << "clear -> removing all elements" << endl;
    l.clear();
    cout << "Size after clear -> " << l.size() << endl;
    cout << "Empty after clear -> " << l.empty() << endl;
    return 0;
}
````
