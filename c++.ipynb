**Calculationns for statistics**

```python
%%writefile calculate.cpp
#include <iostream>
#include <string>
#include <cmath> // Required for pow() and sqrt()

using namespace std;

int main() {
    int a;
    int b;

    // Getting user inputs
    cout << "Enter value a: ";
    cin >> a;
    cout << "Enter value b: ";
    cin >> b;

    // Math Operations
    int multiply = a * b;
    int add = a + b;
    int subtract = a - b;
    int reminder = a % b;


    int div = a / b;
    double fdivision = (double)a / b;

    // Square Roots
    double sqrt_a = sqrt(a);
    double sqrt_b = sqrt(b);

    // Printing Outputs beautifully
    cout << "\n--- RESULTS ---" << endl;
    cout << "Addition (a + b):          " << add << endl;
    cout << "Subtraction (a - b):       " << subtract << endl;
    cout << "Multiplication (a * b):    " << multiply << endl;
    cout << "Integer Division (a / b):  " << div << endl;
    cout << "Decimal Division (a / b):  " << fdivision << endl;
    cout << "Remainder/Modulus (a % b): " << reminder << endl;
    cout << "Square root of a:          " << sqrt_a << endl;
    cout << "Square root of b:          " << sqrt_b << endl;

    // power where we can directly bring ouput
    cout<< "power of a         "<<pow(a,b)<<endl;

    return 0;
}
```

References and its cahnges


```python
%%writefile reference.cpp
#include<iostream>
#include<string>
using namespace std;
int main()
{
  int a = 10;
  int& b = a;
  cout<<a<<endl;
  cout<<&b<<endl;
  a = 20;
  cout<<a<<endl;
  cout<<b<<endl;//here b wont change
  b = 30;
  cout<<a<<endl;// here a will also change becuase its interconnected due to reference
  cout<<b<<endl;
  cout<<"--------------------------"<<endl;
  int health = 100;
  int& playerHealth = health;
  cout<<health<<endl;
  cout<<playerHealth<<endl;
  playerHealth = playerHealth - 20;
  cout<<playerHealth<<endl;
  cout<<health<<endl;
  return 0;
}
```

```python
%%writefile dumb_dynamic_array.cpp
#include <iostream>
using namespace std;

int main() {
    int size;
    cout << "Enter size: ";
    cin >> size;              // user decides size

    int* arr = new int[size]; // create dynamic array

    // Fill array
    for (int i = 0; i < size; i++) {
        arr[i] = i + 1;       // put values 1,2,3,...
    }

    // Print array
    for (int i = 0; i < size; i++) {
        cout << arr[i] << " ";
    }

    delete[] arr;             // free memory
    return 0;
}
```
